```python
import librosa
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plotMelSpectrogram(waveFilename: str, ax = None):
    data, fs = librosa.load(waveFilename)
    print("> Loaded sound from", waveFilename)
    
    L = len(data)
    print("> Read sound with length:", L / fs)

    frameLength = 0.025
    frameSize = int(frameLength * fs)
    print("> NFFT:", frameSize)

    mel_spect = librosa.feature.melspectrogram(y=data, sr=fs, n_mels=128, fmax=8000)
    mel_spect = librosa.power_to_db(mel_spect, ref=np.max)

    if ax is None:
        fig, ax = plt.subplots()
        img = librosa.display.specshow(mel_spect, sr=fs, x_axis='time', y_axis='mel', ax=ax)
        ax.set(title='Mel-frequency spectrogram')
        fig.colorbar(img, ax=ax, format='%+2.0f dB')
        return fig
    else:
        return librosa.display.specshow(mel_spect, sr=fs, x_axis='time', y_axis='mel', ax=ax)
```

```python
fig = plt.figure(figsize=(800, 400), dpi=10)
ax1 = fig.add_subplot(121)
img = plotMelSpectrogram("../MockingBird/wavs/temp_source.wav", ax1)
fig.colorbar(img, ax=ax1, format='%+2.0f dB')
ax2 = fig.add_subplot(122)
img = plotMelSpectrogram("../MockingBird/wavs/temp_result.wav", ax2)
fig.colorbar(img, ax=ax2, format='%+2.0f dB')
fig.show()
```
